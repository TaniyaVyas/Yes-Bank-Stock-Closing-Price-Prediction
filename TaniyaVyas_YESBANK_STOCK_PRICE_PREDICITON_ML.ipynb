## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Create a copy of the original dataset to keep it unchanged
df = df.copy()

# Remove unnecessary spaces from column names
df.columns = df.columns.str.strip()

# Check that all required columns are available
required_columns = ['Date', 'Open', 'High', 'Low', 'Close']

missing_columns = set(required_columns) - set(df.columns)

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Convert the Date column into datetime format
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%b-%y',
    errors='raise'
)

# Convert stock price columns into numerical format
price_columns = ['Open', 'High', 'Low', 'Close']

for column in price_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors='raise'
    )

# Remove exact duplicate records, if present
df.drop_duplicates(inplace=True)

# Sort the dataset chronologically
df.sort_values(by='Date', inplace=True)

# Reset the index after sorting
df.reset_index(drop=True, inplace=True)

# Check for missing values after data conversion
if df.isnull().sum().sum() > 0:
    raise ValueError("Missing values detected after preprocessing.")

# Validate that stock prices are positive
if (df[price_columns] <= 0).any().any():
    raise ValueError("Invalid stock prices detected.")

# Check for duplicate monthly records
if df['Date'].duplicated().any():
    raise ValueError("Duplicate monthly dates detected.")

# Check that High and Low values are consistent
if (df['High'] < df['Low']).any():
    raise ValueError("High price cannot be lower than Low price.")

# Validate that Open and Close are within the monthly range
if (
    (df[['Open', 'Close']].gt(df['High'], axis=0)) |
    (df[['Open', 'Close']].lt(df['Low'], axis=0))
).any().any():
    raise ValueError("Stock prices fall outside the High-Low range.")

# Create Year and Month features for time-based analysis
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Calculate the monthly stock price range
df['Price_Range'] = df['High'] - df['Low']

# Calculate the monthly price change
df['Price_Change'] = df['Close'] - df['Open']

# Calculate percentage change from opening to closing price
df['Price_Change_Percentage'] = (
    (df['Close'] - df['Open']) / df['Open']
) * 100

# Calculate month-to-month closing price returns
df['Monthly_Return'] = df['Close'].pct_change() * 100

# Calculate moving averages for trend analysis
df['MA_3'] = df['Close'].rolling(window=3).mean()
df['MA_6'] = df['Close'].rolling(window=6).mean()

# Display the cleaned dataset
print("Dataset successfully prepared for EDA!")

print("\nFinal Dataset Shape:", df.shape)

print("\nMissing Values in Original Columns:")
print(df[required_columns].isnull().sum())

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()
````

````python
# Chart - 3 visualization code

# Chart 3: Box Plot of Yes Bank Closing Stock Prices

# Create a figure for visualization
plt.figure(figsize=(12, 6))

# Create a box plot to examine the distribution and outliers
sns.boxplot(
    x=df['Close'],
    color='skyblue',
    width=0.4,
    linewidth=1.5,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markersize': 6,
        'markeredgecolor': 'black'
    }
)

# Calculate quartiles and interquartile range
Q1 = df['Close'].quantile(0.25)
Q3 = df['Close'].quantile(0.75)
IQR = Q3 - Q1

# Calculate the lower and upper outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = df[
    (df['Close'] < lower_bound) |
    (df['Close'] > upper_bound)
]

# Add chart title and axis labels
plt.title(
    'Box Plot of Yes Bank Closing Stock Prices',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Closing Stock Price', fontsize=12)

# Add grid lines for readability
plt.grid(axis='x', linestyle='--', alpha=0.4)

# Improve chart layout
plt.tight_layout()

# Display the chart
plt.show()

# Display statistical information
print(f"First Quartile (Q1): {Q1:.2f}")
print(f"Median: {df['Close'].median():.2f}")
print(f"Third Quartile (Q3): {Q3:.2f}")
print(f"Interquartile Range (IQR): {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
print(f"Total Potential Outliers: {len(outliers)}")
````

H_0: \mu_{\text{Close}} = \mu_{\text{Open}}
\]

Alternative Hypothesis (H₁): There is a significant difference between the average monthly opening and closing stock prices of Yes Bank.
\[
H_1: \mu_{\text{Close}} \neq \mu_{\text{Open}}
\]

Significance Level: α = 0.05
2. Perform an Appropriate Statistical Test
We will use the Paired Sample t-test to compare the opening and closing stock prices for the same months.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Hypothesis Testing 2: Paired Sample t-test

# Select opening and closing prices
hypo2_data = df[['Open', 'Close']].dropna()

# Perform paired sample t-test
t_statistic, p_value = ttest_rel(
    hypo2_data['Close'],
    hypo2_data['Open']
)

# Display statistical results
print(f"Mean Opening Price: {hypo2_data['Open'].mean():.2f}")
print(f"Mean Closing Price: {hypo2_data['Close'].mean():.2f}")
print(f"T-Statistic: {t_statistic:.4f}")
print(f"P-Value: {p_value:.6g}")
print("Significance Level: 0.05")

# Interpret the hypothesis test
if p_value < 0.05:
    print("Reject the Null Hypothesis (H0).")
    print("There is a significant difference between opening and closing prices.")
else:
    print("Fail to Reject the Null Hypothesis (H0).")
    print("There is no statistically significant evidence of a difference between opening and closing prices.")
````

````python

# ML Model - 3 Implementation with Hyperparameter Optimization
# Lasso Regression using GridSearchCV

# Create pipeline
lasso_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', Lasso(max_iter=50000))
])

# Define hyperparameter grid
lasso_param_grid = {
    'lasso__alpha': [
        0.001, 0.01, 0.05, 0.1, 0.5,
        1.0, 2.0, 5.0, 10.0, 20.0, 50.0
    ]
}

# Define time-series cross-validation
lasso_tscv = TimeSeriesSplit(n_splits=5)

# Initialize GridSearchCV
lasso_grid_search = GridSearchCV(
    estimator=lasso_pipeline,
    param_grid=lasso_param_grid,
    cv=lasso_tscv,
    scoring='neg_mean_squared_error',
    n_jobs=-1,
    refit=True
)

# Fit the Algorithm
lasso_grid_search.fit(X_train, y_train)

# Get best model
lasso_tuned_model = lasso_grid_search.best_estimator_

# Predict on the model
y_train_pred_lasso_tuned = lasso_tuned_model.predict(X_train)
y_test_pred_lasso_tuned = lasso_tuned_model.predict(X_test)

# Display optimization results
print("Best Hyperparameters:")
print(lasso_grid_search.best_params_)

print("\nBest Cross-Validation RMSE:")
print(np.sqrt(-lasso_grid_search.best_score_))

print("\nLasso Regression optimization completed successfully.")
````

##### Which hyperparameter optimization technique have you used and why?

I used GridSearchCV with five-fold TimeSeriesSplit to optimize the Lasso Regression model.

I selected GridSearchCV because Lasso Regression has one main hyperparameter, alpha, and testing a predefined range of values is computationally manageable for this dataset.

The alpha parameter controls the strength of L1 regularization. Smaller values apply weaker regularization, while larger values shrink coefficients more strongly and may eliminate less useful features.

I used TimeSeriesSplit because the dataset contains monthly stock prices, and preserving chronological order is important for realistic forecasting.

I also used a Pipeline containing StandardScaler and Lasso Regression to prevent data leakage during cross-validation.

I selected the best alpha value based on the lowest cross-validation mean squared error.

Business Impact: A higher R² Score indicates better explanatory performance on the evaluated dataset, which can support historical stock price analysis. However, it does not guarantee accurate future predictions or profitable trading decisions.
