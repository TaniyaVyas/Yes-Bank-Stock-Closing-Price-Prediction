## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Create a copy of the original dataset to keep it unchanged
df = df.copy()

# Remove unnecessary spaces from column names
df.columns = df.columns.str.strip()

# Check that all required columns are available
required_columns = ['Date', 'Open', 'High', 'Low', 'Close']

missing_columns = set(required_columns) - set(df.columns)

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Convert the Date column into datetime format
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%b-%y',
    errors='raise'
)

# Convert stock price columns into numerical format
price_columns = ['Open', 'High', 'Low', 'Close']

for column in price_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors='raise'
    )

# Remove exact duplicate records, if present
df.drop_duplicates(inplace=True)

# Sort the dataset chronologically
df.sort_values(by='Date', inplace=True)

# Reset the index after sorting
df.reset_index(drop=True, inplace=True)

# Check for missing values after data conversion
if df.isnull().sum().sum() > 0:
    raise ValueError("Missing values detected after preprocessing.")

# Validate that stock prices are positive
if (df[price_columns] <= 0).any().any():
    raise ValueError("Invalid stock prices detected.")

# Check for duplicate monthly records
if df['Date'].duplicated().any():
    raise ValueError("Duplicate monthly dates detected.")

# Check that High and Low values are consistent
if (df['High'] < df['Low']).any():
    raise ValueError("High price cannot be lower than Low price.")

# Validate that Open and Close are within the monthly range
if (
    (df[['Open', 'Close']].gt(df['High'], axis=0)) |
    (df[['Open', 'Close']].lt(df['Low'], axis=0))
).any().any():
    raise ValueError("Stock prices fall outside the High-Low range.")

# Create Year and Month features for time-based analysis
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Calculate the monthly stock price range
df['Price_Range'] = df['High'] - df['Low']

# Calculate the monthly price change
df['Price_Change'] = df['Close'] - df['Open']

# Calculate percentage change from opening to closing price
df['Price_Change_Percentage'] = (
    (df['Close'] - df['Open']) / df['Open']
) * 100

# Calculate month-to-month closing price returns
df['Monthly_Return'] = df['Close'].pct_change() * 100

# Calculate moving averages for trend analysis
df['MA_3'] = df['Close'].rolling(window=3).mean()
df['MA_6'] = df['Close'].rolling(window=6).mean()

# Display the cleaned dataset
print("Dataset successfully prepared for EDA!")

print("\nFinal Dataset Shape:", df.shape)

print("\nMissing Values in Original Columns:")
print(df[required_columns].isnull().sum())

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()
````

````python
# Chart - 12 visualization code

# Chart 12: Year-Wise Positive and Negative Monthly Returns

# Sort the dataset chronologically
chart_df = df.sort_values('Date').copy()

# Calculate monthly percentage returns using closing prices
chart_df['Monthly_Return'] = (
    chart_df['Close'].pct_change() * 100
)

# Extract the year from the Date column
chart_df['Year'] = chart_df['Date'].dt.year

# Remove the first undefined monthly return
chart_df = chart_df.dropna(subset=['Monthly_Return'])

# Check whether sufficient data is available
if chart_df.empty:
    raise ValueError("No valid monthly returns available for analysis.")

# Categorize monthly returns as Positive, Negative or Zero
chart_df['Return_Type'] = np.select(
    [
        chart_df['Monthly_Return'] > 0,
        chart_df['Monthly_Return'] < 0
    ],
    [
        'Positive',
        'Negative'
    ],
    default='Zero'
)

# Calculate the average return for each year and return type
yearly_returns = (
    chart_df[chart_df['Return_Type'] != 'Zero']
    .groupby(['Year', 'Return_Type'])['Monthly_Return']
    .mean()
    .unstack()
)

# Ensure both positive and negative return columns exist
yearly_returns = yearly_returns.reindex(
    columns=['Positive', 'Negative']
)

# Prepare yearly positions for plotting
years = sorted(chart_df['Year'].unique())

yearly_returns = yearly_returns.reindex(years)

# Missing categories are left as NaN, not treated as zero returns
positive_returns = yearly_returns['Positive'].values
negative_returns = yearly_returns['Negative'].values

x = np.arange(len(years))
bar_width = 0.38

# Create the grouped bar chart
plt.figure(figsize=(15, 7))

# Plot average positive monthly returns
plt.bar(
    x - bar_width / 2,
    positive_returns,
    width=bar_width,
    color='seagreen',
    label='Average Positive Return'
)

# Plot average negative monthly returns
plt.bar(
    x + bar_width / 2,
    negative_returns,
    width=bar_width,
    color='indianred',
    label='Average Negative Return'
)

# Add a horizontal reference line at zero
plt.axhline(
    y=0,
    color='black',
    linewidth=1
)

# Add chart title and axis labels
plt.title(
    'Year-Wise Average Positive and Negative Monthly Returns',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Monthly Return (%)', fontsize=12)

# Improve chart readability
plt.xticks(x, years, rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.legend()

# Adjust layout and display the chart
plt.tight_layout()
plt.show()

# Display yearly average positive and negative returns
print("Year-Wise Average Positive and Negative Monthly Returns:")
display(yearly_returns.round(2))

# Count positive, negative and zero-return months for each year
return_counts = (
    chart_df.groupby(['Year', 'Return_Type'])
    .size()
    .unstack(fill_value=0)
    .reindex(
        index=years,
        columns=['Positive', 'Negative', 'Zero'],
        fill_value=0
    )
)

print("\nYear-Wise Count of Monthly Return Types:")
display(return_counts)

# Identify the year with the highest average positive return
if yearly_returns['Positive'].notna().any():
    best_positive_year = yearly_returns['Positive'].idxmax()
    print(
        f"\nHighest Average Positive Return Year: "
        f"{best_positive_year}"
    )
    print(
        f"Average Positive Return: "
        f"{yearly_returns.loc[best_positive_year, 'Positive']:.2f}%"
    )

# Identify the year with the most negative average return
if yearly_returns['Negative'].notna().any():
    worst_negative_year = yearly_returns['Negative'].idxmin()
    print(
        f"\nMost Negative Average Return Year: "
        f"{worst_negative_year}"
    )
    print(
        f"Average Negative Return: "
        f"{yearly_returns.loc[worst_negative_year, 'Negative']:.2f}%"
    )
````

````python
# Correlation Heatmap visualization code

# Chart 14: Correlation Heatmap of Numerical Stock Variables

# Sort data chronologically before calculating returns
chart_df = df.sort_values('Date').copy()

# Calculate monthly price range
chart_df['Price_Range'] = (
    chart_df['High'] - chart_df['Low']
)

# Calculate month-to-month percentage returns
chart_df['Monthly_Return'] = (
    chart_df['Close'].pct_change() * 100
)

# Select numerical variables for correlation analysis
numerical_columns = [
    'Open',
    'High',
    'Low',
    'Close',
    'Price_Range',
    'Monthly_Return'
]

# Prepare the numerical dataset
corr_df = chart_df[numerical_columns].copy()

# Replace infinite values with missing values
corr_df = corr_df.replace([np.inf, -np.inf], np.nan)

# Check whether sufficient data is available
if len(corr_df) < 3:
    raise ValueError("Insufficient data for correlation analysis.")

# Calculate the Pearson correlation matrix
correlation_matrix = corr_df.corr(method='pearson')

# Create the figure
plt.figure(figsize=(11, 8))

# Create the correlation heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'Pearson Correlation Coefficient'}
)

# Add chart title
plt.title(
    'Correlation Heatmap of Yes Bank Stock Price Variables',
    fontsize=15,
    fontweight='bold',
    pad=15
)

# Improve chart readability
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

# Adjust layout and display the chart
plt.tight_layout()
plt.show()

# Display the complete correlation matrix
print("Correlation Matrix:")
display(correlation_matrix.round(3))

# Calculate correlations with the target variable
target_correlations = (
    correlation_matrix['Close']
    .drop('Close')
    .sort_values(ascending=False)
)

print("\nCorrelation of Features with Closing Price:")
display(target_correlations.to_frame(name='Correlation_with_Close'))

# Identify highly correlated feature pairs
threshold = 0.90

feature_columns = [
    col for col in numerical_columns if col != 'Close'
]

feature_corr = correlation_matrix.loc[
    feature_columns, feature_columns
]

high_corr_pairs = []

for i in range(len(feature_columns)):
    for j in range(i + 1, len(feature_columns)):

        feature_1 = feature_columns[i]
        feature_2 = feature_columns[j]

        corr_value = feature_corr.loc[feature_1, feature_2]

        if pd.notna(corr_value) and abs(corr_value) >= threshold:
            high_corr_pairs.append({
                'Feature_1': feature_1,
                'Feature_2': feature_2,
                'Correlation': round(corr_value, 3)
            })

# Display highly correlated input feature pairs
print("\nHighly Correlated Feature Pairs (|r| >= 0.90):")

if high_corr_pairs:
    display(pd.DataFrame(high_corr_pairs))
else:
    print("No feature pairs exceeded the selected threshold.")

# Identify the feature with the strongest absolute
# correlation with the target
valid_target_corr = target_correlations.dropna()

if not valid_target_corr.empty:
    strongest_feature = valid_target_corr.abs().idxmax()

    print(
        f"\nStrongest Correlation with Close: "
        f"{strongest_feature}"
    )

    print(
        f"Correlation Value: "
        f"{target_correlations[strongest_feature]:.4f}"
    )
````

````python
# Pair Plot visualization code

# Chart 15: Pair Plot Analysis of Stock Price Variables

# Select the four original numerical stock price columns
price_columns = ['Open', 'High', 'Low', 'Close']

# Prepare a clean dataset for visualization
chart_df = df[price_columns].copy()

# Convert values to numeric and handle invalid entries
for column in price_columns:
    chart_df[column] = pd.to_numeric(
        chart_df[column],
        errors='coerce'
    )

# Remove rows containing missing or infinite values
chart_df = chart_df.replace(
    [np.inf, -np.inf],
    np.nan
).dropna()

# Check whether sufficient observations are available
if len(chart_df) < 3:
    raise ValueError(
        "Insufficient valid observations for pair plot analysis."
    )

# Set the visualization theme
sns.set_theme(style='whitegrid')

# Create a pair plot for Open, High, Low and Close
pair_plot = sns.pairplot(
    data=chart_df,
    vars=price_columns,
    diag_kind='hist',
    kind='scatter',
    plot_kws={
        'color': 'royalblue',
        'alpha': 0.6,
        's': 35,
        'edgecolor': 'white',
        'linewidth': 0.3
    },
    diag_kws={
        'color': 'mediumseagreen',
        'bins': 20,
        'edgecolor': 'white'
    },
    height=2.7
)

# Add an overall title to the pair plot
pair_plot.figure.suptitle(
    'Pair Plot of Yes Bank Stock Price Variables',
    fontsize=16,
    fontweight='bold',
    y=1.02
)

# Improve spacing and display the plot
pair_plot.figure.tight_layout()
plt.show()

# Calculate correlation between the selected variables
correlation_matrix = chart_df.corr(method='pearson')

# Display the correlation matrix
print("Correlation Matrix of Stock Price Variables:")
display(correlation_matrix.round(3))

# Display correlations with the target variable
print("\nCorrelation of Features with Closing Price:")

target_correlations = (
    correlation_matrix['Close']
    .drop('Close')
    .sort_values(ascending=False)
)

display(
    target_correlations.to_frame(
        name='Correlation_with_Close'
    ).round(3)
)

# Identify the strongest relationship with Close
valid_correlations = target_correlations.dropna()

if not valid_correlations.empty:
    strongest_feature = valid_correlations.abs().idxmax()

    print(
        f"\nFeature Most Strongly Correlated with Close: "
        f"{strongest_feature}"
    )

    print(
        f"Correlation Value: "
        f"{target_correlations[strongest_feature]:.4f}"
    )

# Display the number of observations used
print(f"\nTotal Observations Analyzed: {len(chart_df)}")
````
